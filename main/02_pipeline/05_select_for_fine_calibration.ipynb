# 05 Select For Fine Calibration (отбор для тонкой калибровки)

Этот notebook (ноутбук) автоматически отбирает стратегии для `fine mode (тонкого режима)` после завершения `coarse calibration (грубой калибровки)`.

Он не пересчитывает `two-option filter (двухопционный фильтр)`. Он читает готовый файл `03_calibrated_best_by_strategy.csv`, строит `leaderboard (таблицу лидеров)` и сохраняет `02_selected_for_fine.csv`.

Mean reversion (возврат к среднему) отбирается отдельно: часть стратегий может не пройти coarse calibration (грубую калибровку) только из-за промаха по `T (экспирации)` или `center/scale (центру/масштабу)`, поэтому для них сохраняется отдельный `rescue set (спасательный набор)`.

## Критерии

Стратегии ранжируются внутри групп:

- `put/call (пут/колл)`;
- `call/call (колл/колл)`;
- `call/put (колл/пут)`;
- `put/put (пут/пут)`;
- `mean_reversion (возврат к среднему)`.

Критерии: положительный результат на `validation (валидации)`, положительное `improvement (улучшение)`, приемлемый `drawdown (просадка)`, `Sharpe ratio (коэффициент Шарпа)`, `Sortino ratio (коэффициент Сортино)`, и устойчивость формы.

Для mean reversion (возврата к среднему) дополнительно сохраняется `rescue set (спасательный набор)`: стратегии с `half-life (периодом полураспада)`, устойчивой формой или почти окупаемой `validation (валидацией)`, даже если текущие coarse hyperparameters (грубые гиперпараметры) не дали сильной прибыли.

```python
from pathlib import Path
import subprocess
import sys

WORK_ROOT = Path(r"./workspace")
SCRIPT = WORK_ROOT / "select_for_fine_calibration.py"

cmd = [sys.executable, str(SCRIPT), "--mode", "select"]
result = subprocess.run(cmd, capture_output=True, text=True)
print(result.stdout)
print(result.stderr)
```
