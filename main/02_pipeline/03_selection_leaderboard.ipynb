# 03 Selection Leaderboard (таблица отбора)

Этот notebook (ноутбук) строит `leaderboard (таблицу лидеров)` после `two-option screening (первичного двухопционного отбора)`.

Логика: мы не отбрасываем стратегии сразу. Мы ранжируем их внутри групп: `core_only (только базовая стратегия)`, `call/call (колл/колл)`, `put/call (пут/колл)`, `call/put (колл/пут)`, `put/put (пут/пут)`.

## Критерии ранжирования

Порядок критериев:

1. достаточно данных;
2. устойчивость формы;
3. improvement (улучшение) относительно `core strategy (базовой стратегии)`;
4. итоговый `PnL (прибыль и убыток)`;
5. `max drawdown (максимальная просадка)`;
6. `Sharpe ratio (коэффициент Шарпа)`;
7. `Sortino ratio (коэффициент Сортино)`.

Если `core strategy (базовая стратегия)` сама положительна, а фильтр улучшает слабо или просадка фильтра сопоставима с результатом, стратегия помечается как `core_only (только базовая стратегия)`.

```python
from pathlib import Path
import subprocess
import sys

WORK_ROOT = Path(r"./workspace")
SCRIPT = WORK_ROOT / "selection_leaderboard.py"

cmd = [sys.executable, str(SCRIPT), "--mode", "build"]
result = subprocess.run(cmd, capture_output=True, text=True)
print(result.stdout)
print(result.stderr)
```
